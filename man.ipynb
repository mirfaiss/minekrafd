```python
import pandas as pd

# 1. Load data dari spreadsheet (bisa diekspor ke csv/excel dulu)
df_direktori = pd.read_excel('direktori_usaha.xlsx', sheet_name='direktori_usaha')
df_keyword = pd.read_excel('direktori_usaha.xlsx', sheet_name='keyword')

# 2. Bersihkan dan gabungkan keyword menjadi pola regex (contoh: 'foto|orgen|cell')
keywords_bersih = df_keyword['keyword'].dropna().astype(str)
pola_regex = '|'.join(keywords_bersih)

# 3. Filter dataframe (case=False agar tidak sensitif huruf besar/kecil, na=False untuk handle cell kosong)
hasil_filter = df_direktori[df_direktori['nama_usaha'].str.contains(pola_regex, case=False, na=False, regex=True)]

# 4. Simpan hasilnya
hasil_filter.to_excel('hasil_filter_usaha.xlsx', index=False)
```

```python
import pandas as pd

# 1. Load data dari spreadsheet
df_direktori = pd.read_excel('direktori_usaha.xlsx', sheet_name='direktori_usaha')
df_keyword = pd.read_excel('direktori_usaha.xlsx', sheet_name='keyword')

# 2. Bersihkan keyword dari spasi kosong dan nilai kosong
keywords_bersih = df_keyword['keyword'].dropna().astype(str).str.strip()

# 3. Gabungkan menjadi pola regex DENGAN Word Boundary (\b)
# Hasilnya akan seperti: \b(foto|orgen|nikah|studio|...|cell)\b
pola_regex = r'\b(' + '|'.join(keywords_bersih) + r')\b'

# 4. Filter dataframe (case=False agar tidak sensitif huruf besar/kecil)
hasil_filter = df_direktori[df_direktori['nama_usaha'].str.contains(pola_regex, case=False, na=False, regex=True)]

# 5. Simpan hasilnya
hasil_filter.to_excel('hasil_filter_usaha_exact.xlsx', index=False)
```
